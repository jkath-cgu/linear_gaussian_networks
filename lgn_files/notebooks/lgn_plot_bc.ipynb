```python
def find_nearest(array, value):
    array = np.asarray(array)
    idx = (np.abs(array - value)).argmin()
    return idx, array.flatten()[idx]

pct_lhs_set = 0      #
pct_rhs_set = 100    # 100 99
z_max_std = 3

data_z_l_fil = data_z_l[0:len(y_log_hf)][np.array(data_z_l[0:len(y_log_hf)].flatten() <= z_max_std) & np.array(data_z_h.flatten() <= z_max_std)]
data_z_h_fil = data_z_h[np.array(data_z_l[0:len(y_log_hf)].flatten() <= z_max_std) & np.array(data_z_h.flatten() <= z_max_std)]

data_z_h_mean = np.mean(data_z_h_fil)
data_z_h_median = np.median(data_z_h_fil)
data_z_h_std_lhs = np.percentile(data_z_h_fil, pct_lhs_set)
data_z_h_std_rhs = np.percentile(data_z_h_fil, pct_rhs_set)
print(data_z_h_mean, data_z_h_median, data_z_h_std_lhs, data_z_h_std_rhs)

data_z_l_mean = np.mean(data_z_l_fil)
data_z_l_median = np.median(data_z_l_fil)
data_z_l_std_lhs = np.percentile(data_z_l_fil, pct_lhs_set)
data_z_l_std_rhs = np.percentile(data_z_l_fil, pct_rhs_set)
print(data_z_l_mean, data_z_l_median, data_z_l_std_lhs, data_z_l_std_rhs)
```
